```python
import pandas as pd
import numpy as np
```

```python
data_set = pd.read_csv('C:\\Users\\shaik\\PycharmProjects\\ML learning\\data sets\\heart-disease.csv')
```

```python
from sklearn.model_selection import train_test_split
```

```python
data_set.isna().sum()
```

```python
X = data_set.drop('target', axis = 1)
y = data_set['target']
```

```python
data_set.head(5)
```

```python
X_train,X_test, y_train,y_test = train_test_split(X, y, test_size = 0.25, random_state = 42)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
```

```python
clfs = {
    'Kneighbors': KNeighborsClassifier(n_neighbors=100),
    
    'ensemble':RandomForestClassifier(n_estimators = 150),

    'logisticregressor': LogisticRegression(max_iter = 1000)
}
```

```python
for value, model in clfs.items():
    model.fit(X_train, y_train)
    scores = model.score(X_test, y_test)
    print(f'{value}: {scores * 100:.2f}%')
```

```python
from sklearn.model_selection import RandomizedSearchCV, GridSearchCV
```

```python
est = {
    'RandomForest':RandomForestClassifier(),
    'LogisticRegression':LogisticRegression(),
    'Kneighbors': KNeighborsClassifier()
}
Randomgrid = {
    'n_estimators':[100,150,200],
    'criterion':["gini", "entropy", "log_loss"],
    'max_depth':[10,15],
    'min_samples_split':[2],
    'min_samples_leaf':[2,4],
    'max_features':["sqrt", "log2", None]
}
loggrid = {
    'penalty':['l2', 'l1', None],
    'C':[1.0],
    'l1_ratio':[0.0],
    'dual':[False],
    'tol':[0.0001],
    'fit_intercept':[True],
    'intercept_scaling':[1],
}

kgrid = {
    'n_neighbors':[5,10],
    'weights':['uniform','distance'],
    'algorithm':['auto','ball_tree', 'kd_tree', 'brute'],
    'leaf_size':[30],
    'p':[2],
    'metric':['minkowski']
}

grid_names = {
    'RandomForest':Randomgrid,
    'Kneighbors':kgrid,
    'LogisticRegression':loggrid
}
```

```python
for value, model in est.items():
        g_model = RandomizedSearchCV(estimator = model, param_distributions = grid_names[value])
        g_model.fit(X_train, y_train)
        print(f'{value}:  {g_model.score(X_test, y_test) * 100:.2f}%')
```

```python
grids = {
    'RandomForest': Randomgrid,
    'LogisticRegression': loggrid,
    'Kneighbors': kgrid
}

results = {}
for name, model in est.items():
    search = RandomizedSearchCV(estimator=model, param_distributions=grids[name], cv=5)
    search.fit(X_train, y_train)
    results[name] = search.score(X_test, y_test)
    print(search.best_params_)
    print(name, results[name])
    
```

```python
train_score = []
test_score = []

neighbors = range(1,21)
knn = KNeighborsClassifier()

for i in neighbors:
    knn.set_params(n_neighbors = i)
    knn.fit(X_train, y_train)
    train_score.append(knn.score(X_train, y_train))
    test_score.append(knn.score(X_test, y_test))
```
